```python
import matplotlib.pyplot as plt

for label in (t_scope.columns):
    plt.hist(t_scope[t_scope['class']=='g'][label],color='orange',label='Gama',alpha=0.7,density=True)
    plt.hist(t_scope[t_scope['class']=='h'][label],color='green',label='Hadron',alpha=0.7,density=True)

    plt.xlabel(label)
    plt.ylabel("Probability")
    plt.title(label)
    plt.legend()
    plt.show();
```

```python
from sklearn.model_selection import train_test_split
data_train,data_test=train_test_split(t_scope,test_size=0.2)
```

```python
print(len(data_train))
print(len(data_test))
```

```python
print (len(data_train[data_train['class']=='g']))
print(len(data_train[data_train['class']=='h']))
```

```python
from imblearn.over_sampling import RandomOverSampler 
random=RandomOverSampler()
input_data_train=data_train.drop(columns=['class'])
output_data_train=data_train['class']
```

```python
input_data_train
```

```python
output_data_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
input_data_train=scaler.fit_transform(input_data_train)
print(input_data_train)
```

```python
input_data_train,output_data_train=random.fit_resample(input_data_train,output_data_train)
print(len(output_data_train))
print(sum(output_data_train=='g'))
print(sum(output_data_train=='h'))
```

```python
input_data_test=data_test.drop(columns = ['class'])
output_data_test=data_test['class']
input_data_test=scaler.fit_transform(input_data_test )
```

```python
print(input_data_test)
```

```python
print(len(input_data_test))
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(input_data_train,output_data_train)
```

```python
prediction=knn_model.predict(input_data_test)
print(prediction)
```

```python
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(output_data_test,prediction)
accuracy
```
